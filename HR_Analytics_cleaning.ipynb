- EmpID: معرف الموظف الفريد.
- Age: عمر الموظف.
- AgeGroup: الفئة العمرية التي ينتمي إليها الموظف.
- Attrition: ما إذا كان الموظف قد ترك العمل أم لا (Yes/No).
- BusinessTravel: طبيعة السفر المتعلقة بالعمل (Travel_Rarely، Travel_Frequently، Non-Travel).
- DailyRate: المعدل اليومي للراتب.
- Department: القسم الذي يعمل به الموظف (Sales، Research & Development، Human Resources).
- DistanceFromHome: المسافة من المنزل إلى مكان العمل (بالكيلومترات).
- Education: مستوى التعليم (1: أقل تعليم، 5: أعلى تعليم).
- EducationField: مجال الدراسة (Life Sciences، Medical، Marketing، ...).
- EmployeeCount: عدد الموظفين (ثابت لكل الموظفين، قد يكون لعرض بعض الإحصائيات).
- EmployeeNumber: رقم الموظف الفريد (يختلف عن EmpID، قد يكون لأغراض الترتيب).
- EnvironmentSatisfaction: درجة رضا الموظف عن بيئة العمل (1 إلى 4).
- Gender: الجنس (Male/Female).
- HourlyRate: الأجر بالساعة.
- JobInvolvement: درجة مشاركة الموظف في العمل (1 إلى 4).
- JobLevel: المستوى الوظيفي للموظف (1 إلى 5).
- JobRole: الدور الوظيفي للموظف (Sales Executive، Research Scientist، ...).
- JobSatisfaction: رضا الموظف عن وظيفته (1 إلى 4).
- MaritalStatus: الحالة الاجتماعية للموظف (Single، Married، Divorced).
- MonthlyIncome: الدخل الشهري للموظف.
- SalarySlab: شريحة الراتب الشهري.
- MonthlyRate: المعدل الشهري للراتب.
- NumCompaniesWorked: عدد الشركات التي عمل بها الموظف سابقًا.
- Over18: ما إذا كان الموظف فوق 18 عامًا (Yes).
- OverTime: ما إذا كان الموظف يعمل ساعات إضافية (Yes/No).
- PercentSalaryHike: نسبة زيادة الراتب.
- PerformanceRating: تقييم الأداء (1 إلى 4).
- RelationshipSatisfaction: رضا الموظف عن علاقاته في العمل (1 إلى 4).
- StandardHours: عدد ساعات العمل القياسية (ثابت لجميع الموظفين).
- StockOptionLevel: مستوى خيارات الأسهم الممنوحة للموظف (0 إلى 3).
- TotalWorkingYears: إجمالي عدد سنوات العمل.
- TrainingTimesLastYear: عدد مرات التدريب التي حصل عليها الموظف في العام الماضي.
- WorkLifeBalance: توازن الحياة العملية والشخصية (1 إلى 4).
- YearsAtCompany: عدد سنوات العمل في الشركة الحالية.
- YearsInCurrentRole: عدد السنوات في الدور الحالي.
- YearsSinceLastPromotion: عدد السنوات منذ آخر ترقية.
- YearsWithCurrManager: عدد السنوات التي قضاها الموظف مع المدير الحالي.
